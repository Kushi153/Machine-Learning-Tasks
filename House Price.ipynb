### Checking the Outliers

```python
numeric_cols = df.select_dtypes(include="number").columns
print(numeric_cols)
print(len(numeric_cols))

fig, axes = plt.subplots(2, 9, figsize=(20, 8))

axes = axes.flatten()

for i, column in enumerate(numeric_cols):
    sns.boxplot(y=df[column], ax=axes[i])
    axes[i].set_title(column)

# Hide the 18th empty space
axes[17].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
numeric_cols = df.select_dtypes(include="number").columns

# Loop through every numerical column
for column in numeric_cols:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    # Calculate IQR
    IQR = Q3 - Q1

    # Calculate lower and upper limits
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    # Clip the outliers
    df[column] = df[column].clip(lower=lower, upper=upper)

print("Outliers handled using clipping.")
```

# Decision tree  (Regressor)

```python
# decisoon tree model
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error,root_mean_squared_error
dt = DecisionTreeRegressor(random_state=42,max_depth=6,min_samples_leaf=10)

dt.fit(X_train,y_train)

y_pred = dt.predict(X_test)

mse = mean_squared_error(y_test,y_pred)
rmse = root_mean_squared_error(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred)
r2 = r2_score(y_test,y_pred)

print(f"MSE = {mse}\nRMSE = {rmse}\n MAE ={mae}\nR^2 = {r2}")
```

```python
# RandomizedSearchCV
from sklearn.model_selection import RandomizedSearchCV
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error,root_mean_squared_error
dt = DecisionTreeRegressor(random_state=42)

param_dist = {
    "max_depth": [3, 4, 5, 6],
    "min_samples_leaf": [5, 10, 15, 20],
    "min_samples_split": [2, 5, 10],
}

random_search = RandomizedSearchCV(dt, param_distributions=param_dist, n_iter=10, cv=5, random_state=42)
random_search.fit(X_train, y_train)

best_dt = random_search.best_estimator_ 

y_pred = best_dt.predict(X_test)

mse = mean_squared_error(y_test,y_pred)
rmse = root_mean_squared_error(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred)
r2 = r2_score(y_test,y_pred)

print(f"MSE = {mse}\nRMSE = {rmse}\n MAE ={mae}\nR^2 = {r2}")
```

# Random Forest Regressor

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score

# 1. Initialize the Random Forest Regressor
# n_estimators is the number of trees; random_state ensures reproducibility
rf_regressor = RandomForestRegressor(n_estimators=100, max_depth=15, random_state=42, n_jobs=-1)

# 2. Train the model on your training data
rf_regressor.fit(X_train, y_train)

# 3. Predict on the test set
y_pred_rf = rf_regressor.predict(X_test)

# 4. Evaluate the new performance
print("--- Random Forest Results ---")
print("MSE:", mean_squared_error(y_test, y_pred_rf))
print("RMSE:", mean_squared_error(y_test, y_pred_rf))
print("R2 Score:", r2_score(y_test, y_pred_rf))
```

```python
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error, r2_score

# 1. Initialize the XGBoost Regressor
# n_estimators controls the number of trees, learning_rate controls step size shrinkage
xgb_regressor = XGBRegressor(
    n_estimators=100, 
    learning_rate=0.1, 
    max_depth=6, 
    random_state=42, 
    n_jobs=-1
)

# 2. Train the model on your training data
xgb_regressor.fit(X_train, y_train)

# 3. Predict on the test set
y_pred_xgb = xgb_regressor.predict(X_test)

# 4. Evaluate the performance
print("--- XGBoost Results ---")
print("MSE:", mean_squared_error(y_test, y_pred_xgb))
print("RMSE:", mean_squared_error(y_test, y_pred_xgb))
print("R2 Score:", r2_score(y_test, y_pred_xgb))
```

```python
from sklearn.model_selection import RandomizedSearchCV

# 1. Define the hyperparameter grid to search
param_grid = {
    'n_estimators': [100, 200, 500],
    'learning_rate': [0.01, 0.05, 0.1, 0.2],
    'max_depth': [3, 5, 7, 10],
    'subsample': [0.7, 0.8, 1.0],
    'colsample_bytree': [0.7, 0.8, 1.0]
}

# 2. Initialize Randomized Search with Cross-Validation
random_search = RandomizedSearchCV(
    estimator=XGBRegressor(random_state=42, n_jobs=-1),
    param_distributions=param_grid,
    n_iter=10,             # Number of random combinations to try
    scoring='r2',          # Optimize for R2 score
    cv=3,                  # 3-fold cross-validation
    verbose=1,
    random_state=42
)

# 3. Fit it to your training data
random_search.fit(X_train, y_train)

# 4. View the best parameters and best score
print("Best Parameters:", random_search.best_params_)
print("Best R2 Score from Tuning:", random_search.best_score_)
```
